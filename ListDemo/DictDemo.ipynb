```python
Car_M1 ={"Camry": 25000, "Corolla": 20000, "RAV4": 30000, "Highlander": 40000}
Car_M2= {"Civic": 22000, "Accord": 28000, "CR-V": 32000, "Pilot": 35000}
Car_M3 = {"Model S": 80000, "Model 3": 40000, "Model X": 90000, "Model Y": 50000}
dCar = {"Toyota": Car_M1, "Honda": Car_M2, "Tesla": Car_M3}
```

```python
##print(dCar)
for x in dCar:
    print("**")
    print(x)
    for y in dCar[x]:
        print(y, dCar[x][y])
```
