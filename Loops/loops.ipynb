```python
M1={"Camry": 2020}
M2={"Altima": 2021}
M3={"Odyssey": 2022}

Car = {"Toyota": M1, "Nissan": M2, "Honda": M3}
iCarCount = Car.__len__()
print(iCarCount)
for k,v in Car.items():
    print("**", k, "**")
    for k1, v1 in v.items():
        print("Year of ", k1, " is ", v1)
```
