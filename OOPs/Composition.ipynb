```python
class Bird:
    def fly(self):
        print("Bird Flies")
        
class Eat:
    def eat(self):
        print("Can Eat")
        
class Swim:
    def swim(self):
        print("Can swim")
        
class Penguin(Bird):
    def __init__(self, oEat, oSwim):
        self._Eat = oEat
        self._Swim = oSwim
        
class Crow(Bird):
    def __init__(self, oEat):
        self._Eat = oEat
        
oEat1 = Eat()
oSwm1 = Swim()
oPeng = Penguin(oEat1, oSwm1)
oPeng.fly()
oPeng._Eat.eat()
oPeng._Swim.swim()

oCrow = Crow(oEat1)
oCrow.fly()
oCrow._Eat.eat();

```
