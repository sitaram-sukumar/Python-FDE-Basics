```python
class Character:
    def __init__(self, _name, _health):
        self.name = _name
        self.health = _health
        
    def take_damage(self, amount):
        self.health -= amount
    
    def stats(self):
        print(f"{self.name} has {self.health} HP/points left.")

player1 = Character("Arnab", 100)
player1.take_damage(20)
player1.stats()
```

```python
class MyClass:
    def __new__(cls, *args, **kwargs):
        print("Inside __new__()")
        instance = super().__new__(cls)     # creating the actual instance
        return instance
    
    def __init__(self, value):
        print("Inside __init__()")
        self.val = value
    
    def CustomPrint(self):
        print(f"Value of object= {self.val}" )
        
## Normal way --- calls new and init intrinsically
oOriginal = MyClass("James")
oOriginal.CustomPrint()

## Explicitly calling new and init
oExplicit = MyClass.__new__(MyClass)
print(hasattr(oExplicit, "val"))

try:
    oExplicit.CustomPrint()
except NotImplementedError:
    print("Exception")
except MemoryError as me:
    print(f"Memory Exception= {me}")
except AttributeError as ae:
    print(f"Attribute Exception= {ae}")
finally:
    print("Finished")

oExplicit.__init__(20)
print(hasattr(oExplicit, "val"))
oExplicit.CustomPrint()
```
