```python
class PayPal:
    def process_payment(self, amount):
        print(f"Processing PayPal payment of ${amount:.2f}")
        
class GPay:
    def process_payment(self, amount):
        print(f"Processing Google Pay payment of ${amount:.2f}")
        
class Zelle:
    def process_payment(self, amount):
        print(f"Processing Zelle payment of ${amount:.2f}")
        
class PaymentProcessor:    
    def executeCheckOut(self, paymentGateway, amount):
        paymentGateway.process_payment(amount)
        
oZelle = Zelle()
oPaymentProcessor = PaymentProcessor()
oPaymentProcessor.executeCheckOut(oZelle, 100.00);

oGpay = GPay()
oPaymentProcessor.executeCheckOut(oGpay, 50.00);
```
