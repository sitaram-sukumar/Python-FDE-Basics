```python
class Animal:
    def __init__(self, legs):
        self.leg_count = legs
    
    def eat(self):
        print("Yes I can eat")
    
    def LegCount(self):
        print(f"Num Legs: {self.leg_count}")

class Dog(Animal):
    def __init__(self, name, legCount):
        self.name = name;
        super().__init__(legCount)
        self.bark()
        
    def bark(self):
        print(f"{self.name} says woof")
        
class Cat(Animal):
    def __init__(self, name, legCount):
        self.name = name;
        super().__init__(legCount)
        self.meow()
        
    def meow(self):
        print(f"{self.name} says meow")
        
oDog = Dog("Caeser",4)
oCat = Cat("Rufus",4)
oDog.bark()
oCat.meow()
oDog.eat()
oDog.LegCount()
```

```python
## Multiple Inheritance
class CarBlueprint:
    def __init__():
        print("CarBlueprint Init")

class Engine(CarBlueprint):
    def start(self):
        return ("Starting Engine")
        
class Electronics(CarBlueprint):
    def start(self):
        return ("Starting Electronics")
    
class Tesla(Engine, Electronics):
    def __init__(self):
        pass

print(Tesla.__mro__)
oTesla = Tesla();
oTesla.start()
#Tesla.__mro__ = (Electronics, Tesla, CarBlueprint, Engine)  ## Cannot change MRO (method resolution order)
oTesla.start()
```

```python
class Meta(type):
    def mro(cls):
        default = super().mro()
        return [default[0],default[3],default[1],default[2],default[4]]

## Multiple Inheritance
class MusicBlueprint:
    def __init__(self):
        print("MusicBlueprint Init")

class StringIns(MusicBlueprint):
    def start(self):
        return ("Starting Engine")
        
class WindInstru(MusicBlueprint):
    def start(self):
        return ("Starting Electronics")
    
class Guitar(StringIns, WindInstru, metaclass=Meta):
    def __init__(self):
        print("Guitar init")
    
class Drum(StringIns, WindInstru):
    def __init__(self):
        print("Drum init")
   
print(Guitar.__mro__)
print(Drum.__mro__)

oGenIns = MusicBlueprint;
oGenIns = StringIns();
```

```python
class Singleton:
    _instance = None
    def __new__(cls, val):
        if cls._instance == None:
            cls._instance = super().__new__(cls)
            
        print(f"Only 1 instance is created. ID= {id(cls._instance)}")
        #cls.__init__(val)
        return cls._instance
    
    def __init__(self, val):
        self.val = val
        
    def customPrint(self):
        print(f"Val on obj is {self.val}")
        
obj1 = Singleton(10);            obj1.customPrint()
obj2 = Singleton("james");       obj2.customPrint()
obj3 = Singleton(3.14);          obj3.customPrint()   
```
